````python
import os
import numpy as np
import librosa
import pandas as pd

base = "/content/drive/MyDrive/SHE_GUARD_SCREAM_DATA/extracted/NIGENS"

categories = {
    "femaleScream": "scream",
    "maleScream": "scream",
    "femaleSpeech": "speech",
    "maleSpeech": "speech"
}

sample_rate = 22050
clip_seconds = 3
target_length = sample_rate * clip_seconds

records = []

for folder_name, label in categories.items():
    folder = os.path.join(base, folder_name)

    for filename in os.listdir(folder):
        if not filename.lower().endswith(".wav"):
            continue

        path = os.path.join(folder, filename)

        try:
            audio, sr = librosa.load(
                path, sr=sample_rate, mono=True
            )

            if len(audio) < target_length:
                audio = np.pad(
                    audio, (0, target_length - len(audio))
                )
            else:
                audio = audio[:target_length]

            mfcc = librosa.feature.mfcc(
                y=audio, sr=sample_rate, n_mfcc=13
            )

            features = np.mean(mfcc, axis=1)

            row = {
                f"mfcc_{i+1}": float(features[i])
                for i in range(13)
            }

            row["label"] = label
            row["filename"] = filename
            row["source_folder"] = folder_name

            records.append(row)

        except Exception as e:
            print("Skipped:", filename, "|", str(e))

df = pd.DataFrame(records)

print("Processing complete!")
print("Total recordings processed:", len(df))
print("\nClass counts:")
print(df["label"].value_counts())
print("\nFirst 5 rows:")
display(df.head())
````

````python
from sklearn.model_selection import train_test_split

feature_columns = [f"mfcc_{i+1}" for i in range(13)]

X = df[feature_columns]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training recordings:", len(X_train))
print("Testing recordings:", len(X_test))

print("\nTraining classes:")
print(y_train.value_counts())

print("\nTesting classes:")
print(y_test.value_counts())
````

````python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

model.fit(X_train, y_train)

print("Bhai, model training complete! 🎉")
print("Classes learned:", model.classes_)
````

````python
from sklearn.metrics import accuracy_score, classification_report

# Test recordings par prediction
y_pred = model.predict(X_test)

# Accuracy calculate karo
accuracy = accuracy_score(y_test, y_pred)

print("Test Accuracy:", round(accuracy * 100, 2), "%")
print("\nDetailed Results:")
print(classification_report(y_test, y_pred))
````

````python
def extract_audio_features(audio, sr=22050):
    target_length = sr * 3

    if len(audio) < target_length:
        audio = np.pad(audio, (0, target_length - len(audio)))
    else:
        audio = audio[:target_length]

    mfcc = librosa.feature.mfcc(
        y=audio,
        sr=sr,
        n_mfcc=13
    )

    features = np.concatenate([
        np.mean(mfcc, axis=1),
        np.std(mfcc, axis=1)
    ])

    return features

test_features = extract_audio_features(audio, sr)

print("Features extracted:", len(test_features))
print("Expected: 26")
````

````python

import os
import numpy as np
import pandas as pd
import librosa

# Experimental settings
SR = 22050
WINDOW_SEC = 1.5
HOP_SEC = 0.75

window_size = int(SR * WINDOW_SEC)
hop_size = int(SR * HOP_SEC)

window_rows = []

for _, row in full_df.iterrows():
    path = os.path.join(
        DATASET_PATH,
        str(row["source_folder"]),
        str(row["filename"])
    )

    audio, _ = librosa.load(path, sr=SR, mono=True)

    if len(audio) == 0:
        continue

    # Generate overlapping windows
    if len(audio) <= window_size:
        starts = [0]
    else:
        starts = list(
            range(0, len(audio) - window_size + 1, hop_size)
        )

        # Limit long recordings to at most 4 windows
        if len(starts) > 4:
            indexes = np.linspace(
                0, len(starts) - 1, 4, dtype=int
            )
            starts = [starts[i] for i in indexes]

    for start in starts:
        segment = audio[start:start + window_size]

        if len(segment) < window_size:
            segment = np.pad(
                segment, (0, window_size - len(segment))
            )

        mfcc = librosa.feature.mfcc(
            y=segment,
            sr=SR,
            n_mfcc=13
        )

        # Mean + standard deviation for each MFCC
        features = np.concatenate([
            np.mean(mfcc, axis=1),
            np.std(mfcc, axis=1)
        ])

        item = {
            f"mfcc_mean_{i+1}": features[i]
            for i in range(13)
        }
        item.update({
            f"mfcc_std_{i+1}": features[i+13]
            for i in range(13)
        })

        item["label"] = (
            "scream" if row["label"] == "scream"
            else "not_scream"
        )
        item["source_folder"] = row["source_folder"]
        item["recording_id"] = path
        item["window_start_sec"] = start / SR

        window_rows.append(item)

window_df = pd.DataFrame(window_rows)

print("Window dataset shape:", window_df.shape)
print("\nWindows by label:")
print(window_df["label"].value_counts())

print("\nOriginal recordings represented:")
print(window_df["recording_id"].nunique())

print("\nWindows by source folder:")
print(window_df.groupby("source_folder").size())
````

# ***SHE-GUARD: Audio-Based Scream Detection Using Machine Learning***


## 1. Project Overview

SHE-GUARD is a proposed IoT-based women's safety system. Its objective is to identify possible distress sounds and use them as one input to an alert system.

The audio machine learning module processes audio, extracts acoustic features, and predicts whether the sound resembles a scream.

The intended system architecture is:

**Microphone → Audio preprocessing → ML model → Distress candidate → Sensor fusion / alert logic → Application notification**

Manual SOS must remain independent of automatic audio detection.

**Current status:** The audio model is an experimental prototype. Backend integration, hardware integration, sensor fusion, and end-to-end alert testing are separate tasks that still need to be completed.

## 4. Audio Preprocessing

The audio was processed using Python libraries including:

- `librosa` for audio loading and feature extraction
- `numpy` for numerical operations
- `pandas` for tabular data
- `scikit-learn` for machine learning
- `joblib` for saving and loading the trained model

The original model used these preprocessing steps:

1. Load audio in mono.
2. Resample audio to 22,050 Hz.
3. Pad or truncate the clip to 3 seconds.
4. Extract 13 MFCCs.
5. Calculate the mean of each MFCC over time.
6. Use the resulting 13 numerical features for classification.

### What are MFCCs?

MFCC stands for Mel-Frequency Cepstral Coefficients.

MFCCs represent aspects of the short-term spectral shape of audio using a frequency scale related to human hearing. They are commonly used in speech and sound classification.

MFCCs do not directly identify an emergency. The model learns patterns associated with the labels in its training data.

---

## 5. Original Three-Class Model

The first model classified audio into three classes:

- `scream`
- `speech`
- `other_sound`

It used a Random Forest classifier.

```python
from sklearn.ensemble import RandomForestClassifier

audio_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"

```

### Explanation of the parameters

- `n_estimators=200`: builds 200 decision trees.
- `random_state=42`: makes the randomized training procedure reproducible.
- `class_weight="balanced"`: adjusts class weights based on class frequencies.

### Initial evaluation

One random train/test split reported approximately 94.6% accuracy.

However, the official NIGENS eight-fold evaluation on the selected 459 recordings reported approximately:

- Overall accuracy: 91%
- Scream precision: 86%

This showed that high overall accuracy did not guarantee good scream detection.

We tested external male and female scream samples. The original model predicted `other_sound` for both.

For one female-scream sample, the estimated scream score was only 35.5%.

This suggested that the original three-second features were not capturing enough useful information for some external screams.

Changing the target to binary classification simplified the task, but it did not solve the missed-scream problem.

## 7. Short-Window Experiments

A possible issue with long audio clips is that a short scream may be mixed with silence or background sound. Averaging features over the entire clip may dilute distinctive patterns.

We therefore investigated shorter audio windows.

### First experiment: 1.5-second windows

We tested overlapping 1.5-second windows using the original three-class model.

Each window was padded to three seconds before feature extraction.

This experiment did not improve the external female-scream sample. The scream scores were approximately 34–35%.

This was only a diagnostic experiment because the original model had been trained on three-second clips.

### Second experiment: 1.5-second windows with new features

We created 1.5-second windows with a 0.5-second hop.

For each window, we extracted:

- 13 MFCC means
- 13 MFCC standard deviations

This produced 26 features per window.

The resulting dataset contained:

- 1,247 windows
- 161 scream-labelled windows
- 1,086 non-scream windows

The Random Forest achieved approximately 94% window-level accuracy and 66% scream recall during grouped cross-validation.

However, every window inherited the label of its original recording. A window might not actually contain the scream even when its recording was labelled `scream`. This introduces weak-label noise.

---

- Window duration: 0.5 seconds
- Hop duration: 0.25 seconds
- Sample rate: 22,050 Hz
- MFCC coefficients: 13
- Features: 13 means + 13 standard deviations
- Total features per window: 26

The hop is shorter than the window, so adjacent windows overlap.

## 9. Threshold Experiments

The model returns an estimated scream score for each window. We tested different thresholds to examine the trade-off between missed screams and false alarms.

| Threshold | Female recordings detected | Non-scream recordings with false alarms |
|---:|---:|---:|
| 0.40 | 45/45 | 100/383 |
| 0.50 | 45/45 | 67/383 |
| 0.60 | 41/45 | 14/383 |
| 0.70 | 39/45 | 6/383 |
| 0.80 | 27/45 | 5/383 |

Increasing the threshold reduced false alarms but also caused more missed female-scream recordings.

We retained 0.60 as an experimental candidate threshold. It is not a validated safety threshold.

### Any-window rule versus two-consecutive-window rule

At threshold 0.60, the experiments reported:

- Any window positive: 41/45 female recordings detected; 14/383 non-scream recordings with false alarms.
- Two consecutive windows positive: 35/45 female recordings detected; 6/383 non-scream recordings with false alarms.

The two-consecutive rule reduced false alarms but missed more female-scream recordings. We retained the any-window rule in the current predictor because it was more sensitive on this dataset.

---

## 10. Held-Out Recording Evaluation

We split the original 459 recordings into training and test groups using an 80/20 split, stratified by source folder, with `random_state=42`.

The split was made at the recording level before selecting windows. This ensured that training and testing recordings did not overlap.

The model was trained on the training recordings and evaluated on test windows using a fixed threshold of 0.60.

### Observed results

| Metric | Result |
|---|---:|
| Window-level accuracy | 90.53% |
| Window-level scream precision | 88% |
| Window-level scream recall | 40% |
| Female-scream recordings detected | 9/9 |
| All scream recordings detected | 13/15 |
| Non-scream recordings with false alarms | 5/77 |
| Non-scream recording-level false-alarm rate | 6.5% |

Recording-level detection means at least one window in the recording crossed the threshold.

### How should these results be interpreted?

The 9/9 female-scream result is encouraging but represents only nine female-scream recordings in the test set. It does not establish 100% real-world detection.

Window-level recall was only 40%, meaning many scream-labelled windows were missed even though at least one window was detected in each of those nine female-scream recordings.

The same held-out split was inspected during development. It should not be described as a pristine, untouched final test set.

## 13. How the Predictor Works

The predictor:

1. Loads the saved Random Forest model.
2. Reads the input audio file.
3. Resamples the audio to 22,050 Hz and converts it to mono.
4. Creates 0.5-second windows with a 0.25-second hop.
5. Extracts 13 MFCC means and 13 MFCC standard deviations for each window.
6. Gets a scream score for each window.
7. Finds the maximum score across all windows.
8. Returns `scream_candidate` if the maximum score is at least 0.60; otherwise, it returns `not_scream`.

The predictor is a file-level inference function. It does not itself send an emergency notification, connect to hardware, or fuse sensor readings.

### Feature-name warning

During testing, scikit-learn displayed:

`X does not have valid feature names`

This occurred because the model was trained with named DataFrame columns, while the predictor passed a NumPy array to `predict_proba`.

It was a warning, not a fatal error. A cleaned predictor should pass a DataFrame with the same 26 feature names used during training.

## 14. Errors Encountered and Fixes

### Error 1: ZIP extraction failure

**Cause:** Some ZIP entries used Deflate64 compression, which Python's standard `zipfile` could not handle in this case.

**Fix:** The dataset was extracted using 7-Zip (`7z`).

### Error 2: External screams classified as `other_sound`

**Cause:** The original model did not reliably identify some external scream examples.

**Action taken:** We tested binary classification and short-window features rather than relying only on aggregate accuracy.

### Error 3: `KeyError: window_start_sec`

**Cause:** The first short-window dataset did not store timestamps.

**Fix:** We regenerated the dataset and explicitly saved the start time for each window. Fresh out-of-fold predictions were calculated before temporal evaluation.

### Error 4: Pandas `FutureWarning`

**Cause:** A shifted boolean column followed by `.fillna(False)` triggered a warning about future dtype behaviour.

**Fix:** The shifted values were explicitly converted to boolean.

### Error 5: Colab variables missing after runtime reset

**Cause:** Python variables exist only in the current runtime unless saved.

**Fix:** Rerun notebook cells in order, reload saved model files, and recreate dataframes as needed.

## 16. Questions Judges May Ask

### Q1. Is 90.5% the scream-detection accuracy?

No. It is overall window-level accuracy. Scream-window recall on that held-out split was 40%. At recording level, nine out of nine female-scream recordings were detected because at least one window crossed the threshold.

### Q2. Why use short windows?

A long clip can average together silence, background noise, and a short scream. Short overlapping windows allow the model to score smaller portions of audio. This helped some recording-level results, but missed windows and false alarms remain.

### Q3. Does the model detect an actual emergency?

No. It detects audio patterns similar to the scream examples used for training. It cannot confirm an emergency or determine intent.

### Q4. Is sensor fusion already implemented?

Not in this notebook. This work covers the audio classifier and its predictor. Combining audio with wearable sensor readings and the alert backend is a separate integration step.

### Q5. Why not simply increase the threshold?

Increasing the threshold reduces false alarms but can cause more missed screams. Threshold selection depends on the trade-off between false alerts and missed detections and must be validated on independent recordings.

### Q6. Why keep manual SOS independent?

Automatic audio detection can fail because of noise, microphone placement, unusual vocal patterns, or model limitations. Manual SOS should remain a separate trigger.
